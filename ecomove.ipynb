### 1. A time-series visualisation to show mobility trends over time

```python
#Create a time series visualisation to show mobility trends over time, grouping by month and year, 
    #and using the Date column as the x-axis and the trips as the y-axis.
import plotly.express as px

df_cleaned['Date'] = pd.to_datetime(df_cleaned['Date'])
df_cleaned['Year'] = df_cleaned['Date'].dt.year
df_cleaned['Month'] = df_cleaned['Date'].dt.month
time_series_df = df_cleaned.groupby(['Year', 'Month'])['Trips'].sum().reset_index()

fig = px.line(
    time_series_df,
    x="Month",
    y="Trips",
    color="Year",
    markers=True,
)

fig.update_layout(
    title="Mobility Trends Over Time",
    xaxis_title="Month",
    yaxis_title="Trips",
)

fig.show()
```

```python
# Create a histogram to show the average delay in minutes for each mode of transport, using the Mode column 
# as the x-axis and the average delay as the y-axis.

# Convert the delay column to numeric
delay = pd.to_numeric(
    df["Average_Delay_Minutes"],
    errors="coerce"
).dropna()

histogram_data = pd.DataFrame({
    "Delay": delay
})

fig = px.histogram(
    histogram_data,
    x="Delay",
    nbins=26,
    title="Distribution of Delays in Minutes",
    labels={
        "Delay": "Delay (minutes)",
        "count": "Frequency"
    },
    template="plotly_white"
)

fig.update_traces(
    marker_color="#44bd32",
    marker_line_color="black",
    marker_line_width=1
)

fig.update_layout(
    xaxis_title="Delay (minutes)",
    yaxis_title="Frequency",
    bargap=0.05,
    height=500
)
```

### 2. A comparison chart to compare cities, countries, route types, or transport modes

```python
# Total trips by route type and transport mode, grouped by city and country, using a bar chart to visualize the data.
grouped_df = (
df_cleaned.groupby(['Route_Type', 'Transport_Mode', 'City', 'Country'], as_index=False)
.agg({'Trips': 'sum'})
)

fig = px.bar(
    grouped_df,
    x="Route_Type",
    y="Trips",
    color="Transport_Mode",
    barmode="group"
)

fig.update_layout(
    xaxis_title="Route Type",
    yaxis_title="Total Trips",
)

fig.update_traces(
    marker_line_color="black", 
    marker_line_width=0.1)

fig.show()
```

### 3. A relationship-based visualisation, such as a scatter plot or correlation heatmap

```python
# Heatmap to show correlations between metrics
import plotly.figure_factory as ff

cols = [
    'Passengers', 'Trips', 'Average_Delay_Minutes', 'On_Time_Percentage',
    'Ticket_Revenue_EUR', 'Operating_Cost_EUR', 'CO2_Saved_KG',
    'Energy_Consumption_KWh', 'Customer_Rating', 'Accessibility_Complaints'
]
cols = [c for c in cols if c in df_cleaned.columns]

corr = df_cleaned[cols].corr(numeric_only=True)

fig = ff.create_annotated_heatmap(
    z=corr.values,
    x=corr.columns.tolist(),
    y=corr.index.tolist(),
    colorscale="YlGnBu",
    annotation_text=[[f"{v:.2f}" for v in row] for row in corr.values],
)

fig.update_layout(
    xaxis_title="",
    yaxis_title="",
    width=900,
    height=650,
    margin=dict(l=60, r=30, t=60, b=50),
)

fig.update_traces(
    hoverinfo="z",
    showscale=True,
    colorbar=dict(title="Correlation"),
)

fig.show()
```

### 4. A distribution-based visualisation, such as a histogram, box plot, or violin plot

```python
# Distribution Plot - Average delay in minutes by Transport Mode
category_col = "Transport_Mode"
value_col = "Average_Delay_Minutes"

fig = px.violin(
    df_cleaned,
    x=category_col,
    y=value_col,
    color=category_col,           
    box=True,                     
    points=False,
)

fig.update_traces(
    meanline_visible=False
)

fig.update_layout(
    title="Violin Plot: Average delay in minutes by Transport Mode",
    xaxis_title="Transport Mode",
    yaxis_title="Average Delay (mins)",
)

fig.update_xaxes(tickangle=45)

fig.show()
```

```python
# Distribution plot - On time split by weather conditions
fig = px.violin(
    df_cleaned,
    x="Transport_Mode",
    y="On_Time_Percentage",
    color="Weather",
    box=True,          # quartile box (like inner="quartile")
    points=False,
    )

fig.update_layout(
    title="Violin Plot: On Time Percentage by Transport Mode (split by Weather)",
    xaxis_title="Transport Mode",
    yaxis_title="On Time Percentage",
    )

fig.update_traces(orientation="v")
fig.update_xaxes(tickangle=45)

fig.show()
```

### 5.	A performance indicator visualisation showing measures such as revenue, cost, delays, satisfaction, demand, or CO2 saved

```python
# KPI Analysis

import plotly.graph_objects as go
import plotly.subplots as sp

df_kpi = df_cleaned[
    [
        "Date",
        "Ticket_Revenue_EUR",
        "Average_Delay_Minutes",
        "Customer_Rating",
        "CO2_Saved_KG",
        "Accessibility_Complaints",
    ]
].copy()
df_kpi["Date"] = pd.to_datetime(df_kpi["Date"], errors="coerce")
df_kpi = df_kpi.dropna(subset=["Date"])

kpi = {
    "Total Revenue (€)": df_kpi["Ticket_Revenue_EUR"].sum(),
    "Avg Delay (min)": df_kpi["Average_Delay_Minutes"].mean(),
    "Avg Customer Rating": df_kpi["Customer_Rating"].mean(),
    "Total CO2 Saved (kg)": df_kpi["CO2_Saved_KG"].sum(),
    "Accessibility Complaints": df_kpi["Accessibility_Complaints"].sum(),
}

fig = sp.make_subplots(
    rows=1,
    cols=5,
    specs=[
        [
            {"type": "indicator"},
            {"type": "indicator"},
            {"type": "indicator"},
            {"type": "indicator"},
            {"type": "indicator"},
        ]
    ],
    horizontal_spacing=0.03,
)

fig.add_trace(
    go.Indicator(
        mode="number",
        value=kpi["Total Revenue (€)"],
        number={"valueformat": ",.0f"},
        title={"text": "Total Revenue (€)", "font": {"size": 14}},
    ),
    row=1,
    col=1,
)

fig.add_trace(
    go.Indicator(
        mode="number",
        value=kpi["Avg Delay (min)"],
        number={"valueformat": ".2f"},
        title={"text": "Avg Delay (min)", "font": {"size": 14}},
    ),
    row=1,
    col=2,
)

fig.add_trace(
    go.Indicator(
        mode="number",
        value=kpi["Avg Customer Rating"],
        number={"valueformat": ".2f"},
        title={"text": "Avg Customer Rating", "font": {"size": 14}},
    ),
    row=1,
    col=3,
)

fig.add_trace(
    go.Indicator(
        mode="number",
        value=kpi["Total CO2 Saved (kg)"],
        number={"valueformat": ",.0f"},
        title={"text": "Total CO2 Saved (kg)", "font": {"size": 14}},
    ),
    row=1,
    col=4,
)

fig.add_trace(
    go.Indicator(
        mode="number",
        value=kpi["Accessibility Complaints"],
        number={"valueformat": ",.0f"},
        title={"text": "Accessibility Complaints", "font": {"size": 14}},
    ),
    row=1,
    col=5,
)

fig.update_layout(
    template="plotly_white",
    height=260,
    margin=dict(l=10, r=10, t=60, b=10),
)

fig.show()
```
